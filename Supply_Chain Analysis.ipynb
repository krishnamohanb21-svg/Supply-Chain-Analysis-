# 7) Outlier identification (IQR method)

```python
outlier_summary = {}

for col in df.select_dtypes(include="number").columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    outliers = df[(df[col] < lower) | (df[col] > upper)]
    outlier_summary[col] = len(outliers)

outlier_df = pd.DataFrame.from_dict(
    outlier_summary, orient="index", columns=["Outlier Count"]
).sort_values("Outlier Count", ascending=False)

outlier_df
```

4. Several numeric columns (e.g. Defect rates, Costs, Revenue generated) contain outlier values, indicating a few SKUs/orders with unusually high cost, defect, or revenue figures that merit special attention.
5. Costs and shipping performance vary meaningfully by transportation mode and carrier, suggesting opportunities for logistics optimization.

```python
top_product = revenue_by_product["sum"].idxmax()
top_product_rev = revenue_by_product["sum"].max()
low_product = revenue_by_product["sum"].idxmin()

top_location = rev_by_location.idxmax()
top_carrier_cost = shipping_by_carrier.idxmax()
low_carrier_cost = shipping_by_carrier.idxmin()
top_transport_cost = cost_by_mode.idxmax()

avg_defect = df["Defect rates"].mean()
fail_rate = (df["Inspection results"] == "Fail").mean() * 100

print(f"Top revenue product: {top_product} ({top_product_rev:,.0f})")
print(f"Lowest revenue product: {low_product}")
print(f"Top revenue location: {top_location}")
print(f"Highest avg shipping cost carrier: {top_carrier_cost}")
print(f"Lowest avg shipping cost carrier: {low_carrier_cost}")
print(f"Highest total cost transport mode: {top_transport_cost}")
print(f"Average defect rate: {avg_defect:.2f}%")
print(f"Inspection failure rate: {fail_rate:.1f}%")
```

## Business Insights

1. **Top-performing product category** — `skincare` generates the highest total revenue, while `cosmetics` lags behind. Product mix and marketing focus should lean toward high performers.

2. **Regional revenue concentration** — `Mumbai` contributes the highest total revenue among all locations, indicating a geographically concentrated demand/supply base.

3. **Shipping carrier cost gap** — `Carrier A` has the highest average shipping cost, while `Carrier B` is the most cost-efficient carrier. Renegotiating rates or shifting volume toward cheaper, reliable carriers could reduce logistics spend.

4. **Transportation mode cost impact** — `Road` contributes the largest share of total logistics costs, suggesting a review of route/mode mix could unlock savings (e.g., shifting non-urgent shipments to cheaper modes).

5. **Quality / defect risk** — The average defect rate across all SKUs is around 2.3%, and roughly 36% of inspected batches failed quality inspection. Suppliers/SKUs with high defect rates should be flagged for quality audits.

6. **Revenue vs. sales volume relationship** — Revenue does not scale purely linearly with units sold across product types, implying pricing strategy differences materially affect revenue outcomes.

## Business Recommendations

- **R1.** Reallocate marketing and inventory investment toward the strongest revenue-generating category (`skincare`), while investigating and improving underperformance in `cosmetics`.

- **R2.** Consolidate shipments with lower-cost, high-performing carriers (e.g., `Carrier B`) and renegotiate contracts with higher-cost carriers such as `Carrier A` to reduce overall shipping expenditure.

- **R3.** Implement stricter supplier quality controls / audits for SKUs and suppliers with above-average defect rates to reduce failed inspections and associated rework/returns costs.

- **R4.** Optimize the transportation mode mix — shift a portion of volume currently on `Road` to lower-cost alternatives where delivery-time requirements allow, to reduce total logistics costs without harming service levels.
